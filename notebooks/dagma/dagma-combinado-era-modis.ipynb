```python
cols_modis = [
    "modis_Optical_Depth_047",
    "modis_Optical_Depth_055",
    "modis_Column_WV",
]

(df.groupby("nombre_est")[cols_modis]
 .apply(lambda x: x.isna().mean() * 100)
 .round(2))

cols_era5 = [
    "era5_temperature_2m",
    "era5_dewpoint_temperature_2m",
    "era5_u_component_of_wind_10m",
    "era5_v_component_of_wind_10m",
    "era5_boundary_layer_height",
    "era5_relative_humidity_850hPa",
    "era5_surface_pressure",
    "era5_total_precipitation",]

df[cols_era5].describe().T.round(3)
df["temperatura_c"] = df["era5_temperature_2m"] - 273.15
df["dewpoint_c"] = df["era5_dewpoint_temperature_2m"] - 273.15
df["precipitacion_mm"] = df["era5_total_precipitation"] * 1000
df[["temperatura_c", "dewpoint_c", "precipitacion_mm"]].describe().T.round(3)
percentiles = df.groupby("tipo_gas")["concentracion"].quantile([0.5, 0.75, 0.9, 0.95, 0.99, 0.995, 1.0]).unstack()
percentiles.round(3)

outliers = []

for gas, temp in df.groupby("tipo_gas"):
    q1 = temp["concentracion"].quantile(0.25)
    q3 = temp["concentracion"].quantile(0.75)
    iqr = q3 - q1
    limite_sup = q3 + 1.5 * iqr
    
    n_out = (temp["concentracion"] > limite_sup).sum()
    
    outliers.append({
        "gas": gas,
        "registros": len(temp),
        "limite_sup_iqr": limite_sup,
        "outliers": n_out,
        "pct_outliers": n_out / len(temp) * 100,
        "max": temp["concentracion"].max(),
    })

pd.DataFrame(outliers).round(3)

plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x="tipo_gas", y="concentracion", hue="tipo_gas", legend=False)
plt.title("Outliers de concentración por contaminante")
plt.xlabel("Gas")
plt.ylabel("Concentración")
plt.tight_layout()
plt.show()
plt.figure(figsize=(14, 6))

sns.boxplot(
    data=df,
    x="nombre_est",
    y="concentracion",
    hue="tipo_gas",
    fliersize=1,
)

plt.title("Distribución de concentración por estación y contaminante")
plt.xlabel("Estación")
plt.ylabel("Concentración")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

df_mensual = (
    df.groupby([pd.Grouper(key="fecha", freq="ME"), "tipo_gas"])
    ["concentracion"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(14, 5))
sns.lineplot(data=df_mensual, x="fecha", y="concentracion", hue="tipo_gas")
plt.title("Concentración media mensual por contaminante")
plt.xlabel("Fecha")
plt.ylabel("Concentración media")
plt.tight_layout()
plt.show()

corr_cols = [
    "concentracion",
    "temperatura_c",
    "dewpoint_c",
    "era5_boundary_layer_height",
    "era5_relative_humidity_850hPa",
    "era5_surface_pressure",
    "precipitacion_mm",
    "modis_Optical_Depth_047",
    "modis_Optical_Depth_055",
    "modis_Column_WV",
]

plt.figure(figsize=(10, 7))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlación entre concentración, ERA5 y MODIS")
plt.tight_layout()
plt.show()
```

```python
coords = (
    df.groupby(["nombre_est", "fuente"])
    .agg(
        latitud=("latitud", "first"),
        longitud=("longitud", "first"),
        altitud=("altitud", "first"),
        registros=("concentracion", "size"),
    )
    .reset_index()
)

coords
```

```python
from math import radians, sin, cos, sqrt, atan2

def distancia_m(lat1, lon1, lat2, lon2):
    R = 6371000
    p1 = radians(lat1)
    p2 = radians(lat2)
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = sin(dlat / 2) ** 2 + cos(p1) * cos(p2) * sin(dlon / 2) ** 2
    return 2 * R * atan2(sqrt(a), sqrt(1 - a))

diferencias = []

for estacion, temp in coords.groupby("nombre_est"):
    
    if len(temp) > 1:
        a = temp.iloc[0]
        b = temp.iloc[1]
        
        diferencias.append({
            "nombre_est": estacion,
            "fuente_a": a["fuente"],
            "fuente_b": b["fuente"],
            "distancia_m": distancia_m(
                a["latitud"], a["longitud"],
                b["latitud"], b["longitud"],
            ),
            "diferencia_altitud_m": abs(a["altitud"] - b["altitud"]),
            "registros_a": a["registros"],
            "registros_b": b["registros"],
        })

pd.DataFrame(diferencias).sort_values("distancia_m", ascending=False).round(2)
```

```python
coords_canonicas = (
    coords.sort_values(["nombre_est", "registros"], ascending=[True, False])
    .groupby("nombre_est")
    .head(1)
    .set_index("nombre_est")
)
coords_canonicas
```

```python
df_limpio = df.copy()

df_limpio["latitud_original"] = df_limpio["latitud"]
df_limpio["longitud_original"] = df_limpio["longitud"]
df_limpio["altitud_original"] = df_limpio["altitud"]

df_limpio["latitud"] = df_limpio["nombre_est"].map(coords_canonicas["latitud"])
df_limpio["longitud"] = df_limpio["nombre_est"].map(coords_canonicas["longitud"])
df_limpio["altitud"] = df_limpio["nombre_est"].map(coords_canonicas["altitud"])

df_limpio.groupby("nombre_est")[["latitud", "longitud", "altitud"]].nunique()
```
