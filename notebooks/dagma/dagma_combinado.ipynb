```python
import pandas as pd
```

```python
df = pd.read_csv("dagma/df_cristian_dagma.csv")
```

```python
df.head(20)
```

```python
df.info()
```

```python
print(f"Desde: {df['fecha'].min()} hasta {df['fecha'].max()}")
```

```python
columnas_categoricas = ['nombre_est', 'tipo_gas', 'fuente', 'estacion_id', 'municipio']
for col in columnas_categoricas:
    if col in df.columns:
        print(f"\nValores únicos en {col}:")
        print(df[col].unique())
```

```python
print("\nValores únicos de Latitud:")
print(df['latitud'].unique())

print("\nValores únicos de Longitud:")
print(df['longitud'].unique())

print("\nValores únicos de Altitud:")
print(df['altitud'].unique())
```

```python
df.describe()
```

```python
df['concentracion'].describe().round(2)
```

```python
print(df[['nombre_est', 'estacion_id']].drop_duplicates().reset_index(drop=True))
```

```python
print(df.groupby('nombre_est')['tipo_gas'].unique())
```

```python
df_zeros = df[df['concentracion'] == 0]
print(df_zeros.groupby('nombre_est')['concentracion'].count())
```

```python
df_con_datos = df[df['concentracion'] > 0]
print(df_con_datos.groupby('nombre_est')['concentracion'].count())
```

```python
print(df.groupby(['nombre_est', 'tipo_gas'])['concentracion'].describe().round(2))
```

```python
df["fecha"] = pd.to_datetime(df["fecha"])

df_cristian_dagma_2021_2025 = df[
    (df["fecha"].dt.year >= 2021) & (df["fecha"].dt.year <= 2025)
].copy()
```

```python
df_cristian_dagma_2021_2025["fecha"] = pd.to_datetime(df_cristian_dagma_2021_2025["fecha"])
print(df_cristian_dagma_2021_2025["fecha"].dt.year.value_counts().sort_index())
```

```python
df_cristian_dagma_2021_2025.to_csv("df_cristian_dagma_2021_2025.csv", index=False)
```

```python
print(df_cristian_dagma_2021_2025.groupby(['nombre_est', 'tipo_gas'])['concentracion'].describe().round(2))
```

```python
df_zeros = df_cristian_dagma_2021_2025[df_cristian_dagma_2021_2025['concentracion'] == 0]
print(df_zeros.groupby('nombre_est')['concentracion'].count())
```

```python
df_con_datos = df_cristian_dagma_2021_2025[df_cristian_dagma_2021_2025['concentracion'] > 0]
print(df_con_datos.groupby('nombre_est')['concentracion'].count())
```
