## DAGMA CRUCE CON EL DE CRISTIAN

```python
df_raw = pd.read_excel('dagma/dagma-cristian.xlsx', header=None)

fila_estaciones = df_raw.iloc[22].ffill().astype(str).tolist()
fila_gases = df_raw.iloc[23].ffill().astype(str).tolist()

nuevas_columnas = []
for est, gas in zip(fila_estaciones, fila_gases):
    if "fecha" in est.lower() or "hora" in est.lower():
        nuevas_columnas.append("fecha")
    else:
        nuevas_columnas.append(f"{est.strip()}||{gas.strip()}")

df_wide = df_raw.iloc[25:].copy()
df_wide.columns = nuevas_columnas

df_wide = df_wide.replace(['NA', 'nan', 'NaN', ' '], np.nan)

df_vertical = pd.melt(
    df_wide, 
    id_vars=['fecha'],                 
    var_name='estacion_y_gas',          
    value_name='concentracion'          
)


df_vertical[['nombre_est', 'tipo_gas']] = df_vertical['estacion_y_gas'].str.split('||', expand=True, regex=False)

df_vertical = df_vertical.drop(columns=['estacion_y_gas'])

df_vertical['fecha'] = df_vertical['fecha'].astype(str)
df_vertical['fecha'] = df_vertical['fecha'].str.replace(' 24:00', ' 00:00', regex=False)
df_vertical['fecha'] = pd.to_datetime(df_vertical['fecha'], format='mixed')

df_vertical = df_vertical[['nombre_est', 'tipo_gas', 'concentracion', 'fecha']]

print(df_vertical.head(10))
```

```python
gases = ['no2', 'o3', 'so2']

for gas in gases:
    print(f"\n============= COMPARACIÓN PARA EL GAS: {gas.upper()} =============")
    
    # Sacamos los estadísticos de cada DF para ese gas específico
    stats_cristian = df_cristian[df_cristian['tipo_gas'] == gas]['concentracion'].describe()
    stats_dagma = df[df['tipo_gas'] == gas]['concentracion'].describe()
    
    # Los unimos en una sola tablita para comparar lado a lado
    comparativa = pd.DataFrame({'DF Cristian': stats_cristian, 'DF Dagma (Original)': stats_dagma})
    print(comparativa.round(2)) # Redondeamos a 2 decimales para que se vea limpio
```

```python
df_cristian.loc[df_cristian['concentracion'] < 0, 'concentracion'] = 0
```

```python
gases = ['no2', 'o3', 'so2']

for gas in gases:
    print(f"\n============= COMPARACIÓN PARA EL GAS: {gas.upper()} =============")
    stats_cristian = df_cristian[df_cristian['tipo_gas'] == gas]['concentracion'].describe()
    stats_dagma = df[df['tipo_gas'] == gas]['concentracion'].describe()
    
    comparativa = pd.DataFrame({'DF Cristian': stats_cristian, 'DF Dagma (Original)': stats_dagma})
    print(comparativa.round(2))
```

LIMPIAR OUTLIERS CRISTIAN

```python
# Para NO2
q1_no2_c = df_cristian[df_cristian['tipo_gas'] == 'no2']['concentracion'].quantile(0.25)
q3_no2_c = df_cristian[df_cristian['tipo_gas'] == 'no2']['concentracion'].quantile(0.75)
iqr_no2_c = q3_no2_c - q1_no2_c
lim_no2_c = q3_no2_c + 1.5 * iqr_no2_c

# Para O3
q1_o3_c = df_cristian[df_cristian['tipo_gas'] == 'o3']['concentracion'].quantile(0.25)
q3_o3_c = df_cristian[df_cristian['tipo_gas'] == 'o3']['concentracion'].quantile(0.75)
iqr_o3_c = q3_o3_c - q1_o3_c
lim_o3_c = q3_o3_c + 1.5 * iqr_o3_c

# Para SO2
q1_so2_c = df_cristian[df_cristian['tipo_gas'] == 'so2']['concentracion'].quantile(0.25)
q3_so2_c = df_cristian[df_cristian['tipo_gas'] == 'so2']['concentracion'].quantile(0.75)
iqr_so2_c = q3_so2_c - q1_so2_c
lim_so2_c = q3_so2_c + 1.5 * iqr_so2_c

# Aplicar los filtros combinados en df_cristian
df_cristian = df_cristian[
    ((df_cristian['tipo_gas'] == 'no2') & (df_cristian['concentracion'] <= lim_no2_c)) |
    ((df_cristian['tipo_gas'] == 'o3') & (df_cristian['concentracion'] <= lim_o3_c)) |
    ((df_cristian['tipo_gas'] == 'so2') & (df_cristian['concentracion'] <= lim_so2_c))
].copy()
```

LIMPIAR OUTLIERS DAGMA

```python
# Para NO2
q1_no2_d = df[df['tipo_gas'] == 'no2']['concentracion'].quantile(0.25)
q3_no2_d = df[df['tipo_gas'] == 'no2']['concentracion'].quantile(0.75)
iqr_no2_d = q3_no2_d - q1_no2_d
lim_no2_d = q3_no2_d + 1.5 * iqr_no2_d

# Para O3
q1_o3_d = df[df['tipo_gas'] == 'o3']['concentracion'].quantile(0.25)
q3_o3_d = df[df['tipo_gas'] == 'o3']['concentracion'].quantile(0.75)
iqr_o3_d = q3_o3_d - q1_o3_d
lim_o3_d = q3_o3_d + 1.5 * iqr_o3_d

# Para SO2
q1_so2_d = df[df['tipo_gas'] == 'so2']['concentracion'].quantile(0.25)
q3_so2_d = df[df['tipo_gas'] == 'so2']['concentracion'].quantile(0.75)
iqr_so2_d = q3_so2_d - q1_so2_d
lim_so2_d = q3_so2_d + 1.5 * iqr_so2_d

# Aplicar los filtros combinados en df
df = df[
    ((df['tipo_gas'] == 'no2') & (df['concentracion'] <= lim_no2_d)) |
    ((df['tipo_gas'] == 'o3') & (df['concentracion'] <= lim_o3_d)) |
    ((df['tipo_gas'] == 'so2') & (df['concentracion'] <= lim_so2_d))
].copy()
```
